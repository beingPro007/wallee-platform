```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('/home/rdxtreme/walee_image_processing/images/car.jpeg')
```

```python
img
```

```python
print("Shape  :", img.shape)      # (Height, Width, Channels)
print("Height :", img.shape[0])
print("Width  :", img.shape[1])
print("Channels:", img.shape[2])

print("Dtype  :", img.dtype)      # should be uint8

print("Max pixel value:", img.max())    # should be 255
print("Min pixel value:", img.min())    # should be 0

print("Pixel at (0,0)    :", img[0, 0])       # top-left corner
print("Pixel at (100,100):", img[100, 100])   # somewhere in middle

h, w = img.shape[0], img.shape[1]
ratio = w / h
print(f"\nWidth/Height ratio: {ratio:.2f}")

if ratio > 1:
    print("Orientation: LANDSCAPE (screen/monitor format)")
elif ratio < 1:
    print("Orientation: PORTRAIT (mobile format)")
else:
    print("Orientation: SQUARE")
```

```python
img_rgb   = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_gray  = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_hsv   = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# --- Check shapes ---
print("BGR shape  :", img.shape)       # (H, W, 3)
print("RGB shape  :", img_rgb.shape)   # (H, W, 3)  same but channels swapped
print("Gray shape :", img_gray.shape)  # (H, W)     no channel dimension!
print("HSV shape  :", img_hsv.shape)   # (H, W, 3)

# --- Single pixel comparison ---
r, c = 100, 100
print(f"\nSame pixel at (100,100)")
print("BGR :", img[r, c])        # [B, G, R]
print("RGB :", img_rgb[r, c])    # [R, G, B]   — notice values are swapped!
print("Gray:", img_gray[r, c])   # single number
print("HSV :", img_hsv[r, c])    # [H, S, V]

# --- Show all 4 ---
# cv2.imshow("BGR Original", img)
cv2.imshow("Grayscale", img_gray)
# cv2.imshow("HSV", img_hsv)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_gray, cmap='gray')
plt.title("Grayscale")
plt.axis('off')
```

```python
hist = cv2.calcHist([img_gray], [0], None, [256], [0, 256])
```

```python
plt.subplot(1, 3, 3)
plt.plot(hist, color='black')
plt.title("Pixel Intensity Histogram")
plt.xlabel("Pixel Value (0=Black, 255=White)")
plt.ylabel("Pixel Count")
plt.xlim([0, 256])
```

```python
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))
colors = ('b', 'g', 'r')
for i, col in enumerate(colors):
    hist_ch = cv2.calcHist([img], [i], None, [256], [0, 256])
    plt.plot(hist_ch, color=col, label=col.upper())

plt.title("Per Channel Histogram (BGR)")
plt.xlabel("Pixel Value")
plt.ylabel("Count")
plt.legend()
plt.xlim([0, 256])
plt.show()
```
